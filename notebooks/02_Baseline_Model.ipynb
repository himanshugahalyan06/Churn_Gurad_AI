# 02 - Baseline Model (Naive Accuracy)

Most ML projects optimize for accuracy. In a dataset where 74% of customers do not churn, a naive model that predicts 'No Churn' for everyone will achieve 74% accuracy. Let's see this in action.

```python
import pandas as pd
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, classification_report
from churnguard.config import TARGET_COLUMN
from churnguard.data.load import load_data
from churnguard.data.clean import clean_data
from churnguard.data.split import split_data
from churnguard.features.build_features import create_features
from churnguard.preprocessing.pipeline import build_pipeline

# 1. Load and clean
df = load_data()
df = clean_data(df)

# 2. Split
X_train, X_test, y_train, y_test = split_data(df)

# 3. Features and pipeline
X_train_f = create_features(X_train)
X_test_f = create_features(X_test)
pipeline = build_pipeline(X_train_f)

X_train_tf = pipeline.fit_transform(X_train_f)
X_test_tf = pipeline.transform(X_test_f)

# 4. Dummy Classifier (Predicts Majority Class)
dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train_tf, y_train)
y_pred = dummy.predict(X_test_tf)

print("Baseline Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, zero_division=0))
```

As we can see, the model is perfectly 'accurate' ~73% of the time, but completely useless for the business since it catches 0 churners. This is why we need cost-sensitive learning.
